```python
df_criminal = df_criminal.loc[
    ~df_criminal['evento'].isin(['Atendimento pré-hospitalar','Mortes a esclarecer (sem indício de crime)',
        'busca e salvamento','Morte de Agente do Estado','Pessoa Desaparecida','Pessoa Localizada',
        'emissão de alvarás de licença','Morte por intervenção de Agente do Estado',
        'realização de vistorias','Suicídio','Suicídio de Agente do Estado','Arma de Fogo Apreendida'
    ])
]
df_criminal
```

```python
df_crimes_vitimas = df_criminal.loc[
    df_criminal['total_vitima'] >= 0,
    ['uf','evento','data_referencia', 'feminino', 'masculino', 'nao_informado', 'total_vitima']
]
df_crimes_vitimas
```
